```python
import pandas as pd
import plotly.graph_objects as go
```

```python
aps = pd.read_csv("output/ap_positions.csv")
aps
```

```python
traj = pd.read_csv("output/wall_curves.csv")
traj
```

```python
traj.columns
```

```python
aps.columns
```

```python
fig = go.Figure()


for idx in traj['curve_id'].unique():
    subset = traj.loc[traj['curve_id'] == idx]
    fig.add_trace(
        go.Scatter(
            x = subset['x_pdf_pt'],
            y = subset['y_pdf_pt'],
            mode='markers + lines',
            marker=dict({"color" : "black"}),
            showlegend=False
        ),
    )

violet_aps = aps[aps['type'] == 'violet']
for ap_id in violet_aps['id'].unique():
    subset = violet_aps.loc[violet_aps['id'] == ap_id]
    fig.add_trace(
        go.Scatter(
            x = subset['x_pdf_pt'],
            y = subset['y_pdf_pt'],
            mode='markers',
            marker=dict(
                size=8,
                color='rgba(0, 0, 0, 0)',
                line=dict(
                    color='blueviolet',
                    width=2
                )),
            name=str(ap_id)
        ),

    )

orange_aps = aps[aps['type'] == 'orange']
for ap_id in orange_aps['id'].unique():
    subset = orange_aps.loc[orange_aps['id'] == ap_id]
    fig.add_trace(
        go.Scatter(
            x = subset['x_pdf_pt'],
            y = subset['y_pdf_pt'],
            mode='markers',
            marker=dict(
                size=8,
                color='rgba(0, 0, 0, 0)',
                line=dict(
                    color='orange',
                    width=2
                )),
            name=str(ap_id)
        ),

    )

fig.update_layout(
    yaxis=dict(
        autorange='reversed',
        scaleanchor='x',  # Locks the y-scale to the x-scale
        scaleratio=1      # Sets the aspect ratio to 1:1
    )
)
fig.update_layout(width=1500, height=800)
fig.show()
```
